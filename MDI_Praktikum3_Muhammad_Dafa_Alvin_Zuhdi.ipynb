```python
import seaborn as sns

df = sns.load_dataset("titanic")

# Menghitung jumlah missing value per kolom
print(df.isnull().sum())

# Menghitung persentase missing value per kolom
persentase_missing = (df.isnull().sum() / len(df)) * 100
print('\nPersentase missing value per kolom:')
print(persentase_missing[persentase_missing > 0])
```

```python
# Menghitung jumlah baris duplikat
jumlah_duplikat = df.duplicated().sum()
print('Jumlah baris duplikat:', jumlah_duplikat)

# Menampilkan baris yang terduplikasi (jika ada)
df[df.duplicated()]
```

```python
# Memeriksa nilai unik pada kolom kategorikal untuk mendeteksi inkonsistensi penulisan
for col in df.select_dtypes(include='object').columns:
 print(f'{col}: {df[col].unique()}')

# Memeriksa apakah ada nilai numerik yang tidak valid (contoh: usia negatif)
if 'age' in df.columns:
 print('\nJumlah usia tidak valid (< 0):', (df['age'] < 0).sum())

print('Kategori Embarked:', df['embarked'].unique())
```

```python
# Salin dataframe asli agar data mentah tetap tersimpan untuk perbandingan
df_clean = df.copy()

# IMPUTATION: mengisi missing value kolom numerik dengan median
df_clean['age'] = df_clean['age'].fillna(df_clean['age'].median())

# IMPUTATION: mengisi missing value kolom kategorikal dengan modus (nilai paling sering muncul)
df_clean['embarked'] = df_clean['embarked'].fillna(df_clean['embarked'].mode()[0])

# DROP: menghapus kolom dengan proporsi missing value sangat tinggi (contoh: 'deck')
if 'deck' in df_clean.columns:
 df_clean = df_clean.drop(columns=['deck'])

# Verifikasi hasil cleaning
print('Missing value SEBELUM cleaning:')
print(df.isnull().sum().sum())
print('\nMissing value SESUDAH cleaning:')
print(df_clean.isnull().sum().sum())
```
