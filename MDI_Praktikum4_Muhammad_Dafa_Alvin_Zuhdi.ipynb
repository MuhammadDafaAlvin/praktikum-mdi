```python
import pandas as pd

# EXTRACT: membaca data dari sumber (file CSV)
df_source = pd.read_csv('content/train.csv')

# Ambil subset kolom yang relevan untuk kebutuhan analisis tertentu
kolom_dibutuhkan = ['Survived', 'Pclass', 'Sex', 'Age', 'Fare', 'Embarked']

df_extracted = df_source[kolom_dibutuhkan].copy()
print('Data berhasil di-extract. Jumlah baris:', len(df_extracted))
df_extracted.head()
```

```python
data_etl = df_source[["Name", "Sex", "Age", "Survived"]]

data_etl.head()
```

```python
# TRANSFORM: membersihkan missing value
df_transformed = df_extracted.copy()
df_transformed['Age'] = df_transformed['Age'].fillna(df_transformed['Age'].median())
df_transformed['Embarked'] = df_transformed['Embarked'].fillna(df_transformed['Embarked'].mode()[0])

# TRANSFORM: membuat kolom baru (derived column) - kategori usia
def kategori_usia(usia):
  if usia < 18:
    return 'Anak-anak'
  elif usia < 60:
    return 'Dewasa'
  else:
   return 'Lansia'
df_transformed['kategori_usia'] = df_transformed['Age'].apply(kategori_usia)

# TRANSFORM: standardisasi nama kolom menjadi huruf kapital pada label
df_transformed['Sex'] = df_transformed['Sex'].str.capitalize()
print('Data berhasil ditransformasi.')
df_transformed.head()
```

```python
import pandas as pd

# Mengambil data yang diperlukan dari dataset
data_etl = df_source[["Name", "Sex", "Age", "Survived"]].copy()

# Encode gender
data_etl["Sex"] = data_etl["Sex"].map({
    "male": 1,
    "female": 0
})

# Menambahkan kategori usia
data_etl["Kategori_Usia"] = pd.cut(
    data_etl["Age"],
    bins=[0, 18, 40, 60, 80],
    labels=["Anak", "Dewasa Muda", "Dewasa", "Lansia"]
)

# Menampilkan hasil transformasi
data_etl.head()
```

```python
# LOAD: menyimpan data hasil transformasi ke file CSV baru
df_transformed.to_csv('/content/titanic_hasil_etl.csv', index=False)

print('Data berhasil disimpan ke titanic_hasil_etl.csv')
print('Jumlah baris:', len(df_transformed))
print('Jumlah kolom:', len(df_transformed.columns))
print('Kolom:', list(df_transformed.columns))
```

```python
# LOAD: menyimpan data hasil transformasi ke file CSV baru
data_etl.to_csv("/content/titanic_clean.csv", index=False)

print("Dataset sudah disimpan sebagai titanic_clean.csv")
print("Jumlah baris:", len(data_etl))
print("Jumlah kolom:", len(data_etl.columns))
print("Kolom:", list(data_etl.columns))
```
