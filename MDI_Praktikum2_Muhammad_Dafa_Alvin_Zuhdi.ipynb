```python
import pandas as pd

df = pd.read_csv('content/train.csv')

# Memisahkan kolom numerik dan kategorikal secara otomatis
kolom_numerik = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
kolom_kategorikal = df.select_dtypes(include=['object', 'category']).columns.tolist()

print('Kolom Numerik :', kolom_numerik)
print('Kolom Kategorikal:', kolom_kategorikal)

# Identifikasi manual kolom bertipe teks bebas vs kategorikal (kategori dengan sedikit nilai unik)
for col in kolom_kategorikal:
  print(f'{col}: {df[col].nunique()} nilai unik')
```

```python
import pandas as pd

df = pd.read_csv('content/train.csv')

print("Jumlah kategori Gender:", df['Sex'].unique())
print("Jumlah kategori Embarked:", df['Embarked'].unique())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(7,4))
sns.countplot(x='Embarked', data=df)
plt.title('Distribusi Penumpang berdasarkan Pelabuhan Keberangkatan')
plt.xlabel('Pelabuhan')
plt.ylabel('Jumlah Penumpang')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='Sex', data=df)
plt.title('Jumlah Penumpang Berdasarkan Jenis Kelamin')
plt.show()
```
